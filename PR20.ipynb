## Step 1: Import the libraries

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(10)
print("Libraries imported successfully!")
```

## 1. Line Plot

**Data:** Monthly sales of a shop over 12 months.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales_2025 = [20, 22, 25, 30, 28, 35, 40, 38, 33, 30, 45, 50]
sales_2026 = [25, 27, 30, 34, 36, 42, 48, 46, 40, 39, 55, 60]

plt.figure(figsize=(9, 5))
plt.plot(months, sales_2025, color="blue", marker="o", linestyle="-", linewidth=2, label="2025")
plt.plot(months, sales_2026, color="green", marker="s", linestyle="--", linewidth=2, label="2026")

plt.title("Monthly Sales Trend", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Sales (in thousands)")
plt.legend()
plt.grid(True, linestyle=":", alpha=0.7)
plt.show()
```

## 2. Bar Chart

**Data:** Number of students who chose each programming language.

```python
languages = ["Python", "Java", "C++", "JavaScript", "PHP"]
students = [45, 30, 25, 40, 15]
colors = ["gold", "tomato", "skyblue", "limegreen", "orchid"]

plt.figure(figsize=(8, 5))
bars = plt.bar(languages, students, color=colors, edgecolor="black")

# write the value on top of each bar
for bar, value in zip(bars, students):
    plt.text(bar.get_x() + bar.get_width() / 2, value + 0.5, str(value), ha="center")

plt.title("Favourite Programming Language")
plt.xlabel("Language")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Horizontal bar chart and grouped bar chart
x = np.arange(len(languages))
boys = [25, 15, 12, 22, 8]
girls = [20, 15, 13, 18, 7]
width = 0.35

plt.figure(figsize=(8, 5))
plt.bar(x - width / 2, boys, width, label="Boys", color="steelblue")
plt.bar(x + width / 2, girls, width, label="Girls", color="salmon")
plt.xticks(x, languages)
plt.title("Grouped Bar Chart – Language preference by gender")
plt.xlabel("Language")
plt.ylabel("Number of Students")
plt.legend()
plt.show()
```

## 3. Histogram

**Data:** Marks of 200 students (randomly generated, roughly normal).

```python
marks = np.random.normal(loc=65, scale=12, size=200)     # mean=65, std=12
marks = np.clip(marks, 0, 100)                            # keep marks between 0 and 100

plt.figure(figsize=(8, 5))
plt.hist(marks, bins=10, color="mediumpurple", edgecolor="black")
plt.axvline(marks.mean(), color="red", linestyle="--", label=f"Mean = {marks.mean():.1f}")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students (Frequency)")
plt.legend()
plt.show()
```

## 4. Pie Chart

**Data:** Time spent by a student in a day.

```python
activities = ["Study", "Sleep", "Sports", "Social Media", "Others"]
hours = [6, 8, 2, 3, 5]
explode = [0.1, 0, 0, 0, 0]          # pull out the first slice

plt.figure(figsize=(7, 7))
plt.pie(hours, labels=activities, autopct="%1.1f%%", explode=explode,
        startangle=90, shadow=True,
        colors=["#66b3ff", "#99ff99", "#ffcc99", "#ff9999", "#c2c2f0"])
plt.title("How a student spends a day")
plt.show()
```

## 5. Scatter Plot

**Data:** Study hours vs marks obtained (do more hours give more marks?).

```python
study_hours = np.random.uniform(1, 10, 50)
marks_obtained = 35 + 6 * study_hours + np.random.normal(0, 6, 50)   # marks rise with hours + some noise

plt.figure(figsize=(8, 5))
plt.scatter(study_hours, marks_obtained, c=marks_obtained, cmap="viridis",
            s=70, edgecolor="black")
plt.colorbar(label="Marks")

# trend line
m, c = np.polyfit(study_hours, marks_obtained, 1)
plt.plot(study_hours, m * study_hours + c, color="red", linestyle="--", label="Trend line")

plt.title("Study Hours vs Marks")
plt.xlabel("Study hours per day")
plt.ylabel("Marks obtained")
plt.legend()
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()
```

## 6. Subplots – many plots in one figure

Here we place all five plot types together in one figure (2 rows × 3 columns).

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 9))

# (0,0) Line plot
axes[0, 0].plot(months, sales_2026, color="green", marker="o")
axes[0, 0].set_title("Line Plot")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Sales")
axes[0, 0].tick_params(axis="x", rotation=45)

# (0,1) Bar chart
axes[0, 1].bar(languages, students, color=colors, edgecolor="black")
axes[0, 1].set_title("Bar Chart")
axes[0, 1].set_xlabel("Language")
axes[0, 1].set_ylabel("Students")

# (0,2) Histogram
axes[0, 2].hist(marks, bins=10, color="mediumpurple", edgecolor="black")
axes[0, 2].set_title("Histogram")
axes[0, 2].set_xlabel("Marks")
axes[0, 2].set_ylabel("Frequency")

# (1,0) Pie chart
axes[1, 0].pie(hours, labels=activities, autopct="%1.0f%%", startangle=90)
axes[1, 0].set_title("Pie Chart")

# (1,1) Scatter plot
axes[1, 1].scatter(study_hours, marks_obtained, color="teal")
axes[1, 1].set_title("Scatter Plot")
axes[1, 1].set_xlabel("Study hours")
axes[1, 1].set_ylabel("Marks")

# (1,2) Empty axis -> use it for a text note
axes[1, 2].axis("off")
axes[1, 2].text(0.5, 0.5, "Matplotlib\nSubplots Demo", ha="center", va="center",
                fontsize=20, fontweight="bold")

fig.suptitle("Different types of plots using Matplotlib", fontsize=18, fontweight="bold")
plt.tight_layout()
plt.show()
```

## 7. Figure properties and saving the figure

```python
fig, ax = plt.subplots(figsize=(8, 5), facecolor="#f0f0f0")

ax.plot(months, sales_2025, color="darkred", linestyle="-.", marker="D",
        markersize=8, linewidth=2, label="2025 Sales")

ax.set_title("Figure properties demo", fontsize=16, fontweight="bold", color="navy")
ax.set_xlabel("Month", fontsize=12)
ax.set_ylabel("Sales", fontsize=12)
ax.set_ylim(0, 60)                       # y axis limits
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend(loc="upper left")

# Save the figure as an image file
fig.savefig("sales_plot.png", dpi=150, bbox_inches="tight")
plt.show()
print("Figure saved as sales_plot.png")
```

## Conclusion

- **Line plot** shows trends over time.
- **Bar chart** compares categories.
- **Histogram** shows the distribution of data.
- **Pie chart** shows parts of a whole.
- **Scatter plot** shows the relationship between two variables.
- `plt.subplots()` lets us place many plots in one figure, and figure properties (`figsize`, titles, labels, colors, legend, grid, limits) help make the plots clear and professional. Figures can be saved with `savefig()`.
