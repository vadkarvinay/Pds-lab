```python
# Creates and writes data to a text file

f = open("text.txt", "w")

f.write("Hello, World \n")
f.write("Thursday \n")
f.write("Hello, World \n")

f.close()
```

```python
# Read a text file

f = open("text.txt", "r")

data = f.read()
print(data)

f.close()
```

```python
# Append data to text file

f = open("text.txt", "a")

f.write("This is appended data.\n")

f.close()
```

```python
import csv
```

```python
# Creates and writes data to a CSV file

f = open("student.csv", "w", newline="")

writer = csv.writer(f)

writer.writerow(["ID", "Name", "Marks"])
writer.writerow([1, "Darshil", 95])
writer.writerow([2, "Rahul", 88])

f.close()
```

```python
# Read a CSV file

f = open("student.csv", "r")

reader = csv.reader(f)

print(reader)

for row in reader:
    print(row)

f.close()
```

```python
# Append data to CSV file

f = open("student.csv", "a", newline="")

writer = csv.writer(f)

writer.writerow([3, "Priya", 91])

f.close()
```

```python
# Creates and writes data to a Binary EXE file

f = open("binary.exe", "wb")

f.write(b"Hello Python")

f.close()
```

```python
# Read a Binary EXE file

f = open("binary.exe", "rb")

data = f.read()

print(data)

f.close()
```

```python
# Append data to Binary EXE file

f = open("binary.exe", "ab")

f.write(b"\nAppended Data")

f.close()
```
