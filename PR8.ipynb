```python
import pandas as pd

df = pd.read_csv("employee.csv")

print(df.head())
```

```python
hr = df["HourlyRate"]
```

```python
mean = hr.mean()

print("Mean of Hourly Rate =", mean)
```

```python
median = hr.median()

print("Median of Hourly Rate =", median)
```

```python
mode = hr.mode()

print("Mode of Hourly Rate =", mode)
```

```python
variance = hr.var()

print("Variance of Hourly Rate =", variance)
```

```python
std = hr.std()

print("Standard Deviation =", std)
```

```python
minimum = hr.min()
maximum = hr.max()

range_value = maximum - minimum

print("Minimum =", minimum)
print("Maximum =", maximum)
print("Range =", range_value)
```

```python
Q1 = hr.quantile(0.25)
Q2 = hr.quantile(0.50)
Q3 = hr.quantile(0.75)

print("Q1 =", Q1)
print("Q2 =", Q2)
print("Q3 =", Q3)
```

```python
IQR = Q3 - Q1

print("IQR =", IQR)
```
