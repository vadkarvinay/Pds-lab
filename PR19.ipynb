## Step 1: Import the libraries

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

## Step 2: Create the dataset

```python
data = {
    "EmpID":      [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name":       ["Amit", "Riya", "Karan", "Neha", "Raj", "Simran", "Dev", "Anjali", "Rohan", "Pooja"],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR", "Finance", "IT", "HR", "Finance"],
    "Age":        [25, 30, 28, 35, 40, 27, 32, 29, 45, 38],
    "Salary":     [50000, 42000, 55000, 60000, 75000, 40000, 65000, 52000, 48000, 70000],
    "Experience": [2, 5, 3, 10, 15, 2, 8, 4, 20, 12],
}

df = pd.DataFrame(data)
df
```

## 1. Slicing – selecting rows and columns

```python
# Select a single column
print("Single column (Name):")
print(df["Name"].head(3))
```

```python
# Select multiple columns (use double brackets)
df[["Name", "Salary"]]
```

```python
# iloc -> selection by POSITION (row number / column number)
print("First 4 rows using iloc:")
display(df.iloc[0:4])

print("Rows 2 to 5, columns 1 to 3 using iloc:")
display(df.iloc[2:6, 1:4])
```

```python
# loc -> selection by LABEL (index name / column name)
print("Rows 0 to 3 with columns Name and Salary using loc:")
df.loc[0:3, ["Name", "Salary"]]
```

## 2. Filtering – selecting rows by condition

```python
# Employees earning more than 50000
print("Salary > 50000:")
display(df[df["Salary"] > 50000])

# Employees in the IT department
print("Department is IT:")
display(df[df["Department"] == "IT"])
```

```python
# Multiple conditions: use & (AND), | (OR) and put each condition in brackets
print("IT department AND Salary > 50000:")
display(df[(df["Department"] == "IT") & (df["Salary"] > 50000)])

print("HR OR Finance department:")
display(df[(df["Department"] == "HR") | (df["Department"] == "Finance")])
```

```python
# isin() is a shortcut for checking many values
df[df["Department"].isin(["HR", "Finance"]) & (df["Age"] < 35)]
```

## 3. Sorting

```python
# Sort by Salary (lowest to highest)
print("Ascending order of Salary:")
display(df.sort_values(by="Salary"))

# Sort by Salary (highest to lowest)
print("Descending order of Salary:")
display(df.sort_values(by="Salary", ascending=False))
```

```python
# Sort by more than one column: first Department (A-Z), then Salary (high to low)
df.sort_values(by=["Department", "Salary"], ascending=[True, False])
```

## 4. Concatenating – joining DataFrames

```python
# A second DataFrame with new employees (same columns)
new_data = {
    "EmpID":      [111, 112, 113],
    "Name":       ["Kiran", "Meera", "Tarun"],
    "Department": ["IT", "HR", "Finance"],
    "Age":        [26, 31, 36],
    "Salary":     [51000, 46000, 68000],
    "Experience": [3, 6, 11],
}
df_new = pd.DataFrame(new_data)

# Concatenate ROW-wise (stack one below the other)
df_all = pd.concat([df, df_new], ignore_index=True)
print("After row-wise concatenation, shape =", df_all.shape)
df_all
```

```python
# Concatenate COLUMN-wise (side by side)
extra_cols = pd.DataFrame({
    "City": ["Surat", "Vadodara", "Surat", "Rajkot", "Ahmedabad",
             "Surat", "Rajkot", "Vadodara", "Surat", "Ahmedabad"]
})

df_side = pd.concat([df, extra_cols], axis=1)
df_side
```

## 5. Aggregating – summarizing data

```python
# Overall summary of numeric columns
df[["Age", "Salary", "Experience"]].agg(["min", "max", "mean", "sum"])
```

```python
# Group by Department and calculate the average salary
print("Average salary per department:")
df.groupby("Department")["Salary"].mean()
```

```python
# Many aggregations at once
summary = df.groupby("Department").agg(
    Employees=("EmpID", "count"),
    Avg_Salary=("Salary", "mean"),
    Max_Salary=("Salary", "max"),
    Total_Salary=("Salary", "sum"),
    Avg_Experience=("Experience", "mean"),
)
summary
```

## 6. Normalizing – scaling the data

Salary is in thousands while Experience is in single digits. To compare them fairly we scale them to a common range.

**Min-Max Normalization** (range 0 to 1):
x_new = (x − min) / (max − min)

**Z-score Standardization** (mean 0, std 1):
z = (x − mean) / std

```python
df_norm = df.copy()

# ---- Min-Max normalization ----
for col in ["Age", "Salary", "Experience"]:
    df_norm[col + "_minmax"] = (df[col] - df[col].min()) / (df[col].max() - df[col].min())

df_norm[["Name", "Salary", "Salary_minmax", "Experience", "Experience_minmax"]]
```

```python
# ---- Z-score standardization ----
for col in ["Age", "Salary", "Experience"]:
    df_norm[col + "_zscore"] = (df[col] - df[col].mean()) / df[col].std()

df_norm[["Name", "Salary", "Salary_zscore", "Experience", "Experience_zscore"]]
```

```python
# Verify the results
print("Min-Max scaled columns -> min and max:")
print(df_norm[["Age_minmax", "Salary_minmax", "Experience_minmax"]].agg(["min", "max"]))

print("\nZ-score columns -> mean and std:")
print(df_norm[["Age_zscore", "Salary_zscore", "Experience_zscore"]].agg(["mean", "std"]).round(4))
```

## Conclusion

- **Slicing** was done with `[]`, `iloc` and `loc`.
- **Filtering** was done with boolean conditions using `&`, `|` and `isin()`.
- **Sorting** was done with `sort_values()` on one or many columns.
- **Concatenation** was done row-wise and column-wise with `pd.concat()`.
- **Aggregation** was done with `agg()` and `groupby()`.
- **Normalization** was done with Min-Max (0 to 1) and Z-score (mean 0, std 1) methods, bringing all columns to a comparable scale.
