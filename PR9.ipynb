```python
import pandas as pd

df = pd.read_csv("employee.csv")
```

```python
print(df["HourlyRate"].describe())
```

```python
import matplotlib.pyplot as plt

plt.hist(df["HourlyRate"], bins=10)

plt.xlabel("HourlyRate")
plt.ylabel("Number of Employees")
plt.title("Distribution of Employee Age")

plt.show()
```

```python
plt.scatter(df["HourlyRate"], df["MonthlyIncome"])

plt.xlabel("HourlyRate")
plt.ylabel("Monthly Income")
plt.title("HourlyRate vs Monthly Income")

plt.show()
```

```python
correlation = df["HourlyRate"].corr(df["MonthlyIncome"])

print("Correlation =", correlation)
```

```python
variables = ["HourlyRate","MonthlyIncome","YearsAtCompany","TotalWorkingYears"]

correlation_matrix = df[variables].corr()

print(correlation_matrix.round(2))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Employee Variables")
plt.show()
```
