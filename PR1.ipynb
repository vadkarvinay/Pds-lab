```python
c = 15
print("Number C :- ", c)
c += 3
print("After c += 3 : Num C :- ", c)
c -= 5
print("After c -= 5 : Num C :- ", c)
c *= 2
print("After c *= 2 : Num C :- ", c)
c /= 4
print("After c /= 4 : Num C :- ", c)
c //= 2
print("After c //= 2 : Num C :- ", c)
c %= 3
print("After c %= 3 : Num C :- ", c)
c **= 2
print("After c **= 2 : Num C :- ", c)
```
