```python
import numpy as np
from scipy import stats

print("SciPy statistical analysis started successfully.")
```

```python
# 1. Create a dataset

data = np.array([12, 15, 18, 20, 22, 25, 28, 30, 32, 35])

print("Dataset:")
print(data)
print("Number of observations:", len(data))
```

```python
# 2. Statistical Measures

mean = stats.tmean(data)
median = np.median(data)
mode_result = stats.mode(data, keepdims=False)
variance = np.var(data, ddof=1)
std_dev = np.std(data, ddof=1)
minimum = np.min(data)
maximum = np.max(data)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode_result.mode)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
# 3. Percentiles, Skewness and Kurtosis

q25, q50, q75 = np.percentile(data, [25, 50, 75])
skewness = stats.skew(data)
kurtosis = stats.kurtosis(data)

print("25th Percentile:", q25)
print("50th Percentile:", q50)
print("75th Percentile:", q75)
print("Skewness:", skewness)
print("Kurtosis:", kurtosis)
```

```python
# 4. Z-Scores

z_scores = stats.zscore(data)

print("Z-Scores:")
print(np.round(z_scores, 3))
```

```python
# 5. Correlation

x = np.array([10, 20, 30, 40, 50])
y = np.array([12, 24, 29, 41, 52])

correlation = stats.pearsonr(x, y)

print("Pearson correlation coefficient:", round(correlation.statistic, 4))
print("P-value:", round(correlation.pvalue, 4))
```

## Statistical Tests

The following tests are demonstrated:
1. One-sample t-test
2. Independent two-sample t-test
3. Chi-square test
4. Shapiro-Wilk normality test

```python
# 6. One-Sample T-Test

# Test whether the sample mean is significantly different from 25
t_test_one = stats.ttest_1samp(data, popmean=25)

print("One-Sample T-Test")
print("T-statistic:", round(t_test_one.statistic, 4))
print("P-value:", round(t_test_one.pvalue, 4))
```

```python
# 7. Independent Two-Sample T-Test

group_a = np.array([20, 22, 21, 19, 23, 24])
group_b = np.array([25, 27, 26, 28, 24, 29])

t_test_two = stats.ttest_ind(group_a, group_b)

print("Group A:", group_a)
print("Group B:", group_b)
print("T-statistic:", round(t_test_two.statistic, 4))
print("P-value:", round(t_test_two.pvalue, 4))
```

```python
# 8. Chi-Square Test

# Observed frequencies for three categories
observed = np.array([30, 45, 25])

# Expected frequencies
expected = np.array([33.33, 33.33, 33.34])

chi_test = stats.chisquare(observed, f_exp=expected)

print("Observed frequencies:", observed)
print("Expected frequencies:", expected)
print("Chi-square statistic:", round(chi_test.statistic, 4))
print("P-value:", round(chi_test.pvalue, 4))
```

```python
# 9. Shapiro-Wilk Normality Test

normality_test = stats.shapiro(data)

print("Shapiro-Wilk Test")
print("Statistic:", round(normality_test.statistic, 4))
print("P-value:", round(normality_test.pvalue, 4))
```

```python
# 10. Basic interpretation of statistical tests

alpha = 0.05

print("Significance level (alpha):", alpha)

print("\nOne-Sample T-Test:")
if t_test_one.pvalue < alpha:
    print("Reject H0: The sample mean is significantly different from 25.")
else:
    print("Fail to reject H0: No significant difference from 25.")

print("\nIndependent Two-Sample T-Test:")
if t_test_two.pvalue < alpha:
    print("Reject H0: The two group means are significantly different.")
else:
    print("Fail to reject H0: No significant difference between the group means.")

print("\nChi-Square Test:")
if chi_test.pvalue < alpha:
    print("Reject H0: Observed and expected frequencies differ significantly.")
else:
    print("Fail to reject H0: No significant difference between observed and expected frequencies.")

print("\nShapiro-Wilk Test:")
if normality_test.pvalue < alpha:
    print("Reject H0: The data may not follow a normal distribution.")
else:
    print("Fail to reject H0: The data is consistent with a normal distribution.")
```
